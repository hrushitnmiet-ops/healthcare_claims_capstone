```python
count_orphans = len(df_claim_lines[df_claim_lines["claim_number"] == "__ORPHAN__"])

print(f"Number of orphan claim lines: {count_orphans}")
```

```python
duplicates_column_df = df_claim_lines[df_claim_lines.duplicated(subset=["claim_number", "line_id"])]

duplicates_column_count = len(duplicates_column_df)
print(f"Number of duplicate claim_number and line_id rows: {duplicates_column_count}")
```

```python
clean_claim_lines_df = df_claim_lines[df_claim_lines["claim_number"] != "__ORPHAN__"]
```

```python
clean_claim_lines_df.head(10)
```

```python
clean_claim_lines_df = clean_claim_lines_df.drop_duplicates(subset=["line_id", "claim_number"])
```
